# 1. Analyse exploratoire de données (EDA)

## **La distribution de l'âge**

```python
# Créer un histogramme pour la distribution de l'âge
import plotly.express as px

fig_age = px.histogram(data_assurance, x="age", title="Distribution des âges", labels={"age": "Tranche d'âge"})
fig_age.update_layout(bargap=0.1,yaxis_title="Fréquence d'âge") # ajuster l'espacement entre les barres de l'histogramme et le titre de l'axe y
fig_age.show()
```

A partir d'ici nous ne pouvons tirer de conclusion globale donc nous créerons une variable tranche d'âge pour la suite de la distribution

```python
# Creer ue nouvelle variable dans le data_frame 

def define_tranche_age(age):
    if age < 20 :
        return '<20'
    elif age >=20 and age <30 :
        return '20-30' 
    elif age >=30 and age <40 :
        return '30-40' 
    elif age >=40 and age <50 :
        return '40-50' 
    elif age >=50 and age <60 :
        return '50-60'
    else: 
        return '>=60' 

data_assurance["tranche_age"] = data_assurance["age"].apply(define_tranche_age)
```

```python
data_assurance.head(10)
```

```python
data_assurance_grouped_age = data_assurance["tranche_age"].value_counts().reset_index(name="frequence")
```

```python
data_assurance_grouped_age
```

```python
fig_age_2 = px.bar(
    data_assurance_grouped_age, 
    x="tranche_age", 
    y="frequence", 
    title="Distribution des tranches d'âge", 
    labels={"tranche_age": "Tranche d'âge"},
    text_auto=True)
fig_age_2.update_layout(yaxis_title="Fréquence d'âge")
fig_age_2.show()
```

La répartition des tranches d'âges ci-dessus nous montre que les personnes âgées sont plus intéressées aux produits de l'assureur.

# 2. Analyse des sinistres

## Quel type d’assurance génère le plus de sinistres ?

```python
data_assurance_grouped_type_sinistre = data_assurance.groupby("type_assurance")["nb_sinistres"].sum().reset_index(name="total_sinistres")

data_assurance_grouped_type_sinistre
```

```python
fig_sinistres = px.bar(
    data_assurance_grouped_type_sinistre, 
    x="type_assurance", 
    y="total_sinistres", 
    title="Répartition de sinistres par type d'assurance", 
    labels={"type_assurance": "Type d'assurance"},
    )
fig_sinistres.update_layout(yaxis_title="Nombre de sinistres")
fig_sinistres.show()
```

# 3. Analyse de la rentabilité

```python
data_assurance["profit_client"] = data_assurance["montant_prime"] - data_assurance["montant_sinistres"]

data_assurance.head(10)
```

## Clients rentables vs non rentables

```python
data_assurance["rentabilite"] = data_assurance["profit_client"].apply(lambda x: "Rentable" if x > 0 else "Non rentable")

#data_assurance.drop(columns=["is_rentable"], inplace=True)

data_assurance.head(10)

```

```python
data_assurance_grouped_rentable = data_assurance["rentabilite"].value_counts().reset_index(name="nombre_clients")

data_assurance_grouped_rentable
```

```python

fig_rentable = px.bar(
    data_assurance_grouped_rentable, 
    x="rentabilite", 
    y="nombre_clients", 
    title="Répartition de rentabilité par client", 
    labels={"rentabilite": "Rentabilité"},
    text_auto=True,

    )
fig_rentable.update_layout(yaxis_title="Nombre de clients")
fig_rentable.show()
```

## Rentabilité par type d’assurance

```python
data_assurance_grouped_rentable_type = data_assurance.groupby("type_assurance")["rentabilite"].value_counts().reset_index(name="nombre_clients")

data_assurance_grouped_rentable_type
```

```python
fig_rentable_type = px.bar(
    data_assurance_grouped_rentable_type,
    x="type_assurance",
    y="nombre_clients",
    color="rentabilite",   # séparation des barres
    barmode="group",      # barres côte à côte
    title="Répartition de rentabilité par type d'assurance",
    labels={
        "type_assurance": "Type d'assurance",
        "rentabilite": "Rentabilité"
    },
    text_auto=True
)

fig_rentable_type.update_layout(
    yaxis_title="Nombre de clients"
)

fig_rentable_type.show()
```
