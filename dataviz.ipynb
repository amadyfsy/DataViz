distribution de l'age

```python
import plotly.graph_objects as go

fig = go.Figure(
    data=[go.Histogram(x=data_assurance["age"])],
)

fig.update_layout(
    title="Histogramme des âges",
    bargap=0.2  # espace entre les barres (0 à 1)
)

fig.show()
```

```python
# Créer une nouvelle variable dans le DataFrame

def define_tranche_age(age):
    if age < 20:
        return "<20"
    elif 20 <= age < 30:
        return "20-30"
    elif 30 <= age < 40:
        return "30-40"
    elif 40 <= age < 50:
        return "40-50"
    elif 50 <= age < 60:
        return "50-60"
    else:
        return ">=60"

data_assurance["tranche_age"] = data_assurance["age"].apply(define_tranche_age)
data_assurance.head()
```

```python
data_assurance_grouped = data_assurance["tranche_age"].value_counts().reset_index(name="frequence")
data_assurance_grouped
```

```python
fig = go.Figure(
    data=[
        go.Bar(
            x=data_assurance_grouped["tranche_age"],
            y=data_assurance_grouped["frequence"],
            text=data_assurance_grouped["tranche_age"],  # texte au-dessus
            textposition="outside"
        )
    ]
)

fig.update_layout(
    title="Répartition des assurés par tranche d'âge",
    xaxis_title="Tranches d'âge",
    yaxis_title="Fréquence",
    bargap=0.2
)

fig.show()
```

la repartition des tranches d'age si dessou nous montre que les personnes ages sont plus interesse aux produits de l'assureur
pour ce type de graph donner des recommandations
